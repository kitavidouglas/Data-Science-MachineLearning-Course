```python
import numpy as np
import os
import psutil
import time

# --- Memory-Based (The "Naive" Approach) ---
def memory_based_loading(image_paths):
    start_mem = psutil.Process().memory_info().rss / (1024 * 1024)
    images = [some_load_function(p) for p in image_paths] # Loads everything to RAM
    end_mem = psutil.Process().memory_info().rss / (1024 * 1024)
    print(f"Memory Used: {end_mem - start_mem:.2f} MB")
    return np.array(images)

# --- Generator-Based (The "Elite" Approach) ---
def generator_based_loading(image_paths, batch_size=32):
    """
    Yields data in chunks. RAM usage remains constant regardless 
    of total dataset size.
    """
    for i in range(0, len(image_paths), batch_size):
        batch_paths = image_paths[i:i + batch_size]
        batch_images = [some_load_function(p) for p in batch_paths]
        yield np.array(batch_images)

# Insight: Memory-based loading scales O(N) with data size, 
# while Generators scale O(1) relative to total dataset size.
```
